Import Library

```python
from sklearn import datasets
```

Load dataset

```python
diabetes = datasets.load_diabetes()
```

```python
diabetes
```

Description of the Diabetes dataset

```python
print(diabetes.DESCR)
```

Feature names

```python
print(diabetes.feature_names)
```

Create x and y data matricies

```python
x = diabetes.data
y = diabetes.target
```

```python
x.shape, y.shape
```

import library to split data

```python
from sklearn.model_selection import train_test_split
```

perform 80/20 split

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)
```

data dimension

```python
x_train.shape, y_train.shape
```

```python
x_test.shape, y_test.shape
```

Linear Regression Model

Import library

```python
from sklearn import linear_model
from sklearn.metrics import mean_squared_error, r2_score
```

Build Linear Regression Model

Defines Linear Regression model

```python
model = linear_model.LinearRegression()
```

Build Training Model

```python
model.fit(x_train, y_train)
```

Apply trained model to make a prediction (on test set)

```python
y_pred = model.predict(x_test)
```

Prediction Results

Print model performance

```python
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("Mean squared error (MSE): %.2f"
  % mean_squared_error(y_test, y_pred))
print("Coefficient of determination (R^2): %.2f"
  % r2_score(y_test, y_pred))
```

Coefficients represent the weight values of each of the 10 variables

Boston Housing Dataset

```python
import pandas as pd
BostonHousing = pd.read_csv('BostonHousing.csv')
```

```python
BostonHousing
```

```python
y = BostonHousing.medv
y
```

```python
x = BostonHousing.drop(['medv'], axis = 1)
x
```

data split

```python
from sklearn.model_selection import train_test_split
```

perform 80/20 split

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)
```

```python
x_test.shape, y_test.shape
```

import library

```python
from sklearn import linear_model
from sklearn.metrics import mean_squared_error, r2_score
```

build linear regression

```python
model = linear_model.LinearRegression()
```

build training model

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
print('Coefficients:', model.coef_)
print('Intercept:', model.intercept_)
print('Mean squared error (MSE): %.2f'
% mean_squared_error(y_test, y_pred))
print('Coefficient of determination (R^2): %.2f'
% r2_score(y_test, y_pred))
```
